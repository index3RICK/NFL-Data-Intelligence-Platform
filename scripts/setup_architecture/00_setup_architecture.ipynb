# Create Schemas | Creación de esquemas

## Purpose

This notebook initializes the logical structure of the **NFL Data Intelligence Platform** using the **Medallion Architecture**.

The project is organized into three data layers:

* **Bronze**: Raw data ingested from external sources with minimal transformation.
* **Silver**: Cleaned, standardized, and validated data.
* **Gold**: Business-ready data prepared for analytics and downstream applications.

## What this step does

The SQL cell below creates the three schemas required for the Medallion Architecture:

```text
nfl_bronze
nfl_silver
nfl_gold
```

`IF NOT EXISTS` ensures that the schemas are created only when they do not already exist, making the script safe to execute again without unnecessarily recreating the existing structure.

This step only creates the schemas. No source data is ingested and no transformations are performed at this stage.

## Architecture

```text
External Sources
      │
      ▼
  nfl_bronze
      │
      ▼
  nfl_silver
      │
      ▼
   nfl_gold
```

The schemas will be populated later through the project's ingestion and transformation processes.


````python
%sql
-- =====================================================================
-- Create Schemas | NFL Data Intelligence Platform
-- =====================================================================
-- Creates the three schemas used by the Medallion Architecture.
-- IF NOT EXISTS prevents errors if a schema already exists.
-- =====================================================================

-- Bronze Layer
CREATE SCHEMA IF NOT EXISTS nfl_bronze;

-- Silver Layer
CREATE SCHEMA IF NOT EXISTS nfl_silver;

-- Gold Layer
CREATE SCHEMA IF NOT EXISTS nfl_gold;

-- Validate schema creation
SHOW SCHEMAS;
````
